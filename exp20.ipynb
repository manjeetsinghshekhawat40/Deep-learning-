```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

documents = [
    "I love deep learning",
    "Deep learning is powerful",
    "I love Python programming"
]

# ============================================================
# 1. Bag of Words using CountVectorizer
# ============================================================

bow_vectorizer = CountVectorizer()

bow = bow_vectorizer.fit_transform(documents)

bow_df = pd.DataFrame(
    bow.toarray(),
    columns=bow_vectorizer.get_feature_names_out()
)

print("Bag of Words:\n")
print(bow_df)

# ============================================================
# 2. Bag of n-grams using CountVectorizer
# ============================================================

ngram_vectorizer = CountVectorizer(ngram_range=(2,2))

ngrams = ngram_vectorizer.fit_transform(documents)

ngram_df = pd.DataFrame(
    ngrams.toarray(),
    columns=ngram_vectorizer.get_feature_names_out()
)

print("\nBag of n-grams:\n")
print(ngram_df)

# ============================================================
# 3. Bag of Words using TF-IDF Vectorizer
# ============================================================

tfidf_vectorizer = TfidfVectorizer()

tfidf = tfidf_vectorizer.fit_transform(documents)

tfidf_df = pd.DataFrame(
    tfidf.toarray(),
    columns=tfidf_vectorizer.get_feature_names_out()
)

print("\nTF-IDF Representation:\n")
print(tfidf_df)
```
